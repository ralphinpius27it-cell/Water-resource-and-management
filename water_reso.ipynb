```python
import random
import time
import csv
from datetime import datetime
import matplotlib.pyplot as plt

# File to store water usage logs
FILE_NAME = "water_usage.csv"

# Thresholds
MAX_USAGE_PER_MIN = 5       # liters per minute (normal max flow)
LEAK_DETECTION_THRESHOLD = 0.5  # flow when no usage expected (leak)

# Function to simulate sensor readings
def get_water_flow():
    # Random water flow (liters/min) simulation
    # 70% of time = normal use (0-5 L/min), 20% leak, 10% high flow
    rand = random.random()
    if rand < 0.7:
        return round(random.uniform(0, 5), 2)
    elif rand < 0.9:
        return round(random.uniform(0.1, 0.8), 2)  # Leak
    else:
        return round(random.uniform(5, 10), 2)  # High abnormal usage

# Function to log water usage
def log_data(flow_rate):
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    with open(FILE_NAME, "a", newline="") as file:
        writer = csv.writer(file)
        writer.writerow([timestamp, flow_rate])

# Function to check leak or abnormal usage
def check_alert(flow_rate):
    if flow_rate > MAX_USAGE_PER_MIN:
        print(f"⚠️ ALERT: Abnormally high water usage detected! ({flow_rate} L/min)")
    elif 0 < flow_rate < LEAK_DETECTION_THRESHOLD:
        print(f"💧 ALERT: Possible leakage detected! ({flow_rate} L/min)")

# Function to visualize data
def plot_usage():
    timestamps = []
    flow_rates = []

    with open(FILE_NAME, "r") as file:
        reader = csv.reader(file)
        next(reader, None)
        for row in reader:
            if row:  # skip empty rows
                timestamps.append(row[0])
                flow_rates.append(float(row[1]))

    plt.figure(figsize=(10, 5))
    plt.plot(timestamps, flow_rates, marker='o', linestyle='-', label="Water Flow (L/min)")
    plt.xticks(rotation=45)
    plt.xlabel("Time")
    plt.ylabel("Water Flow (L/min)")
    plt.title("Smart Water Monitoring System")
    plt.legend()
    plt.tight_layout()
    plt.show()

# ------------------ MAIN ------------------
print("Starting Smart Water Monitoring System...\n")

# Create CSV with headers if not exists
try:
    with open(FILE_NAME, "x", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["Timestamp", "FlowRate(L/min)"])
except FileExistsError:
    pass

# Simulate readings for 20 cycles
for i in range(20):
    flow = get_water_flow()
    print(f"[{i+1}] Flow Rate: {flow} L/min")
    log_data(flow)
    check_alert(flow)
    time.sleep(1)  # 1 second delay (instead of real 1 min for demo)

print("\nSimulation completed. Generating graph...")
plot_usage()
```
